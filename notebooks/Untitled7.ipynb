```python
from ai_medreview.sheethelper import SheetHelper
from ai_medreview.params import *
from ai_medreview.utils import *
import pandas as pd
```

```python
@time_it
def load_google_sheet():
    sh = SheetHelper(
        sheet_url="https://docs.google.com/spreadsheets/d/1c-811fFJYT9ulCneTZ7Z8b4CK4feEDRheR0Zea5--d0/edit#gid=0",
        sheet_id=0,
    )
    data = sh.gsheet_to_df()
    data.columns = [
        "submission_id",
        "respondent-id",
        "time",
        "rating",
        "free_text",
        "do_better",
        "pcn",
        "surgery",
        "campaing_id",
        "logic",
        "campaign_rating",
        "campaign_freetext",
    ]

    data["time"] = pd.to_datetime(data["time"], format="%Y-%m-%d %H:%M:%S")

    data.sort_values(by="time", inplace=True)
    return data
```

```python
data = load_google_sheet()
```

```python
dates = pd.to_datetime(data["time"], errors="coerce")
october_data = data.loc[
    (dates >= "2026-09-01") & (dates < "2026-10-01")
]

for surgery in surgeries:
    print(surgery)
    my_surgery = october_data[october_data["surgery"] == surgery]
    print(my_surgery['rating'].value_counts())
    print()
```
