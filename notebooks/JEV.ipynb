```python
data = pd.read_csv('/Users/janduplessis/Downloads/Friends & Family Test Tally - Sheet1.csv')
```

```python
data.head()
```

```python
data.columns
```

```python
data.columns = ['id', 'id2', 'date', 'rating', 'review', 'improvement', 'pcn', 'surgery', 'online1', 'online2', 'online3', 'online4']
```

```python
data.head()
```

```python
data['date'] = pd.to_datetime(data['date'], format="%Y-%m-%d %H:%M:%S")
```

```python
new = data[data['date'] > '2026-01-01'].copy()
```

```python
new['date'].min()
```

```python
new['date'].max()
```

```python
new.drop(columns = ['online1', 'online2','online3','online4', 'id', 'id2'], inplace=True)
```

```python
feed = new.dropna(subset='review')
imp = new.dropna(subset='improvement')
```

```python
feed.drop(columns= ['improvement'], inplace=True)
imp.drop(columns= ['review'], inplace=True)

imp.columns = ['date', 'rating', 'review', 'pcn', 'surgery']
```

```python
feed.shape
```

```python
imp.shape
```

```python
combined = pd.concat([feed, imp], axis=0)
```

```python
combined.shape
```

```python
combined.to_csv('/Users/janduplessis/code/janduplessis883/ai-medreview-3/data/new_data.csv', index=False)
```

```python
# Count words in each review
combined["word_count"] = (
    combined["review"]
    .fillna("")
    .astype(str)
    .str.split()
    .str.len()
)

# Keep reviews with at least 6 words
combined = combined[combined["word_count"] >= 6].copy()

# Optional: reset row numbering
combined.reset_index(drop=True, inplace=True)
```
